```python
import numpy as np
from scipy.optimize import least_squares
import mph

# Modules for GUI file selection.
import tkinter as tk
from tkinter import filedialog
import sys

# %% Import COMSOL Model ======================================================

# Start COMSOL server.
client = mph.start(port=2036)
print(f'COMSOL version {client.version}')
print(f'Server port n° {client.port}')
print(f'Number of cores {client.cores}')
print('\n')

# Select file.
root = tk.Tk()
root.withdraw()
file_path = filedialog.askopenfilename(title = 'Select COMSOL file',
                                       filetypes = [("COMSOL file", "*.mph"), ],)
if file_path=='':
    sys.exit()  # Stop execution immediately.

# Import COMSOL model.
pymodel = client.load(file_path)
model = pymodel.java
print(f'COMSOL model {file_path} loaded\n')

# Show studies.
#tag = model.study().tags()[0]
#print(tag)

# %% Solve initial swelling ===================================================
'''
Compute the initial swelling stretches lambda01, lambda02, lambda03 for given
matrix, fibre and isotonic osmotic pressure difference beta0
'''

def getFreeSwellingParameters(model,suffix):
    '''
    ---------------------------------------------------------------------------
    PURPOSE
      Extract the parameters required for the free swelling computation from
      the COMSOL model.
      
    INPUT
          model   : [object] COMSOL model.
          suffix  : [str] Suffix of the parameters names.
          
    OUTPUT
          matParam: [dict, n° parameters] Elastic parameters.
          osmParam: [dict, n° parameters] Osmotic parameters.
          phis    : [npy array, n° of fibers x 1] Fibers in-plane angles.
          thetas  : [npy array, n° of fibers x 1] Fibers polar angles.
          w       : [npy array, n° of fibers x 1] Fibers weights.
          
    NOTE
    ---------------------------------------------------------------------------
    AUTHOR:                                                          R.Alberini
    LAST MODIFIED:                                                   09-09-2025
    ---------------------------------------------------------------------------
    '''
    # Get the required material parameters.
    matParam = {
        'phi_SR': float(model.param().evaluate('phi_SR'+suffix)),
        'mu0'   : float(model.param().evaluate('mu0'   +suffix,'MPa')),
        'q'     : float(model.param().evaluate('q'     +suffix)),
        'm1'    : float(model.param().evaluate('m1'    +suffix)),
        'm2'    : float(model.param().evaluate('m2'    +suffix)),
        'm3_f'  : float(model.param().evaluate('m3_f'  +suffix)),
        'm4_f'  : float(model.param().evaluate('m4_f'  +suffix)),
        }
    
    # Osmotic parameters.
    osmParam = {
        'c_fcR'   : float(model.param().evaluate('c_fcR'+suffix)),
        'cext_ref': float(model.param().evaluate('cext_ref')),
        'RT'      : float(model.param().evaluate('RT')),
        }
    
    # # Find variables tags containing the identifier 'varfiber' (corresponding
    # # to a fiber direction).
    # varsTags = [str(varTag) for varTag in model.variable().tags()]

    # # Extract fiber variables.
    # idxfib = [i for i, varTag in enumerate(varsTags) if 'varfiber' in varTag]
    # varsTagsFib = [varsTags[i] for i in idxfib]
    
    # # Get fiber orientations.
    # phis    = [] # Azimuthal angles.
    # thetas  = [] # Polar angle.
    # weights = [] # Weights for each discrete fiber direction.
    # for i, tag in enumerate(varsTagsFib):
    #     n = i+1
    #     try:
    #         phis.append(float(model.param().evaluate(f'phi{n}')))
    #         thetas.append(float(model.param().evaluate(f'theta{n}')))
    #         weights.append(float(model.param().evaluate(f'ww{n}')))
    #     except:
    #         print(f'Tag {tag} not available')
    
    # Extract list of parameters names from the collection 'parfiber1'.
    parFibersNames = [str(name) for name in model.param('parfiber_RD').varnames()]
    
    # Extract full list of parameters names.
    parFibersNames = [str(name) for name in model.param().varnames()]
    
    # Get fiber orientations and weights.
    phis    = [] # Azimuthal angles.
    thetas  = [] # Polar angle.
    weights = [] # Weights for each discrete fiber direction.
    for parName in parFibersNames:
        if 'phi'+suffix in parName:
            phis.append(float(model.param().evaluate(parName)))
        elif 'theta'+suffix in parName:
            thetas.append(float(model.param().evaluate(parName)))
        elif 'ww'+suffix in parName:
            weights.append(float(model.param().evaluate(parName)))
        # else:
        #     print(f'Parameter {parName} not relevant')
    
    # Show number of fibers.
    print(f'Fiber directions found: {len(phis)} phi angels, {len(thetas)} theta angels, {len(weights)} weights')
    
    # Return variables.
    return matParam, osmParam, np.array(phis), np.array(thetas), np.array(weights)

# Define Cauchy stress tensor components function.
def CauchyComponentsOsmEq(lambdas, phs, ths, w, osmoticParam, materialParam):
    '''
    ---------------------------------------------------------------------------
    PURPOSE
      Compute Cauchy stress components at osmotic equilibrium assuming 
      F_tot = F*F0 = I*F0, mu_f = 0, sigma_tot = 0.
    
    INPUT
          lambdas      : [npy array, 3 x 1] Initial stretches.
          phs          : [npy array, n° of fibers x 1] Fibers in-plane angles.
          ths          : [npy array, n° of fibers x 1] Fibers polar angles.
          w            : [npy array, n° of fibers x 1] Fibers weights.
          osmoticParam : [dict, n° parameters] Osmotic parameters.
          materialParam: [dict, n° parameters] Elastic parameters.
    
    OUTPUT
          stresses     : [npy array, 3 x 1] Stresses at osmotic equilibrium 
                         (they shall be all equal 0 when plugging the initial
                          swelling stretches).
    
    NOTE
    ---------------------------------------------------------------------------
    AUTHOR:                                                          R.Alberini
    LAST MODIFIED:                                                   09-09-2025
    ---------------------------------------------------------------------------
    '''
    
    # Compute osmotic pressure at equilibrium (F_tot = F*F0 = I*F0, mu_f = 0,
    # sigma_tot = 0).
    c_fcR    = osmoticParam['c_fcR'];   # Reference fixed charges concentration.
    cext_ref = osmoticParam['cext_ref'] # External solution ions concentration.
    RT       = osmoticParam['RT']       # Gas constant x temperature.          
    beta0    = RT*(np.sqrt(c_fcR**2+4*cext_ref**2)-2*cext_ref) # Osmotic pressure [MPa].
    
    # Material parameters.
    phi_SR = materialParam['phi_SR']
    mu0    = materialParam['mu0']
    q      = materialParam['q']
    m1     = materialParam['m1']
    m2     = materialParam['m2']
    m3_f   = materialParam['m3_f']
    m4_f   = materialParam['m4_f']
    
    if m4_f==0:
        m4_f = 1
    
    # Assign stretches.
    lam_01, lam_02, lam_03 = lambdas
    
    # Compute squared sine and cosine of azimuth and polar angles of the fibers.
    sin2_ths = np.sin(ths)**2
    cos2_ths = np.cos(ths)**2
    sin2_phs = np.sin(phs)**2
    cos2_phs = np.cos(phs)**2
    
    # Compute fibers stretches and strains.
    stretches_f = ( sin2_ths * ( lam_01**(-2) * cos2_phs + lam_02**(-2) * sin2_phs ) + lam_03**(-2) * cos2_ths )**(-0.5)
    switches_f  = stretches_f >= 1.0
    strains_f   = stretches_f - 1.0
    
    # Compute generalized strains.
    J0 = lam_01*lam_02*lam_03
    g_fe = (m3_f/m4_f) * np.sum( w * switches_f * strains_f**(2*m4_f) )
    g_m  = m1*(lam_01**2 + lam_02**2 + lam_03**2 - 3.0) + (m1/m2)*(J0**(-2*m2) - 1.0)
    g_tot = g_m + g_fe
    
    # Compute Cauchy stress components.
    s11 = ( phi_SR*mu0*np.exp( q*g_tot ) ) * ( m1*(lam_01**2 - J0**(-2*m2))
    + m3_f * np.sum( w * switches_f * stretches_f * strains_f**(2*m4_f-1.0) * sin2_ths * cos2_phs ) ) - beta0
    
    s22 = ( phi_SR*mu0*np.exp( q*g_tot ) ) * ( m1*(lam_02**2 - J0**(-2*m2))
    + m3_f * np.sum( w * switches_f * stretches_f * strains_f**(2*m4_f-1.0) * sin2_ths * sin2_phs ) ) - beta0
                    
    s33 = ( phi_SR*mu0*np.exp( q*g_tot ) ) * ( m1*(lam_03**2 - J0**(-2*m2))
    + m3_f * np.sum( w * switches_f * stretches_f * strains_f**(2*m4_f-1.0) * cos2_ths ) ) - beta0
    
    return np.array([s11, s22, s33])
    

# Prompt for material.
suffixes = ['_RD', '_AT', '_MF']
suffixlist = ''.join(f'\n > {i:3n}: {suff}' for i, suff in enumerate(suffixes))
val = input(f"Select material suffix from the list:{suffixlist}\ninput element number: ")
suffix = suffixes[int(val)]
print(f'Selected material is: {suffix}\n')

# Get the relevant parameters from the model.
matParam, osmParam, phis, thetas, weights = getFreeSwellingParameters(model,suffix)

# Compute swelling stretches (nonlinear least squares).
resultSwelling = least_squares(
    CauchyComponentsOsmEq,
    x0 = np.array([1.0, 1.0, 1.0]),
    bounds = (np.array([1.0, 1.0, 1.0]), np.array([10., 10., 10.])),
    args=(phis, thetas, weights, osmParam, matParam),
    #ftol = 1e-9,
    #xtol = 1e-9,
    gtol = 1e-11, # Tolerance on the function residual.
    tr_solver = 'exact',
    method = 'trf',
    verbose = 0, # (0 (default) : work silently., 1 : display a termination report, 2 : display progress during iterations)
    )

# Print initial swelling stretches.
lam1 = resultSwelling.x[0]
lam2 = resultSwelling.x[1]
lam3 = resultSwelling.x[2]
print(f'\nSwelling stretches are:\nlam01 = {lam1:10.8f}\nlam02 = {lam2:10.8f}\nlam03 = {lam3:10.8f}\n')

# Print residual stresses using the stretches.
rs11 = resultSwelling.fun[0]
rs22 = resultSwelling.fun[1]
rs33 = resultSwelling.fun[2]
print(f'Residual stresses are:\ns01 = {rs11:10.8e}\ns02 = {rs22:10.8e}\ns03 = {rs33:10.8e}\n')



```
